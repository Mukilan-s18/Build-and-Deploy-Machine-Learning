# Experiment 5: Unsupervised Learning Models
- Clustering with K-Means: Applying K-Means clustering on a dataset
- Dimensionality Reduction with PCA: Performing PCA for feature reduction and visualization

## 1. Generate Data

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import os

os.makedirs('outputs', exist_ok=True)

# Generate 4 clusters
X, y_true = make_blobs(n_samples=500, centers=4, cluster_std=1.5, n_features=5, random_state=42)
print(f"Generated dataset with shape: {X.shape}")
```

## 2. K-Means Clustering

```python
from sklearn.cluster import KMeans

# Apply K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(X)

print("K-Means clustering completed.")
```

## 3. Dimensionality Reduction with PCA

```python
from sklearn.decomposition import PCA

# Reduce to 2 dimensions for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print(f"Explained variance ratio: {pca.explained_variance_ratio_}")
print(f"Total explained variance: {sum(pca.explained_variance_ratio_):.4f}")
```

## 4. Visualize Results

```python
plt.figure(figsize=(10, 6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_kmeans, cmap='viridis', s=50, alpha=0.7)
plt.title('K-Means Clustering projected onto 2D PCA space')
plt.xlabel(f'First Principal Component ({pca.explained_variance_ratio_[0]*100:.1f}%)')
plt.ylabel(f'Second Principal Component ({pca.explained_variance_ratio_[1]*100:.1f}%)')
plt.colorbar(scatter, label='Cluster Label')

# Plot cluster centers
centers = kmeans.cluster_centers_
centers_pca = pca.transform(centers)
plt.scatter(centers_pca[:, 0], centers_pca[:, 1], c='red', s=200, alpha=0.8, marker='X', label='Centroids')
plt.legend()

plt.savefig('outputs/kmeans_pca_visualization.png')
plt.close()
print("Visualization saved to outputs/kmeans_pca_visualization.png")
```
