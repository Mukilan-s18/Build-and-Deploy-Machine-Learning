## 1. Generate a Synthetic Dataset
We'll create a synthetic dataset with missing values and categorical features to demonstrate preprocessing.

```python
import pandas as pd
import numpy as np
import os

# Create directory for dataset if it doesn't exist
os.makedirs('data', exist_ok=True)

# Generate synthetic data
np.random.seed(42)
data = {
    'Age': np.random.randint(18, 65, 100).astype(float),
    'Salary': np.random.randint(30000, 120000, 100).astype(float),
    'Department': np.random.choice(['IT', 'HR', 'Finance', 'Marketing'], 100),
    'Purchased': np.random.choice(['Yes', 'No'], 100)
}

# Introduce missing values
data['Age'][np.random.choice(100, 10, replace=False)] = np.nan
data['Salary'][np.random.choice(100, 15, replace=False)] = np.nan

df = pd.DataFrame(data)
df.to_csv('data/messy_dataset.csv', index=False)
print("Synthetic dataset created at data/messy_dataset.csv")
df.head(10)
```

## 2. Data Cleaning and Preprocessing
Now we will load the dataset, handle missing values, encode categorical features, and scale the numerical features.

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Load data
df = pd.read_csv('data/messy_dataset.csv')

# Separate features and target
X = df.drop('Purchased', axis=1)
y = df['Purchased']

# Define numerical and categorical columns
numeric_features = ['Age', 'Salary']
categorical_features = ['Department']

# Create preprocessing pipelines
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combine preprocessors
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

# Preprocess features
X_processed = preprocessor.fit_transform(X)

# Encode target
le = LabelEncoder()
y_encoded = le.fit_transform(y)

print("Preprocessing complete.")
print(f"Original shape: {X.shape}, Processed shape: {X_processed.shape}")
```
