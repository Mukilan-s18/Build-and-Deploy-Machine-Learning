# Experiment 6: Building a Simple Neural Network
- Creating a feedforward neural network using a library (e.g., TensorFlow/Keras)
- Convolutional Neural Networks (CNN): Implementing a CNN for image classification tasks

## 1. Load and Preprocess MNIST Dataset

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt
import os

os.makedirs('outputs', exist_ok=True)

# Load data
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize pixel values
x_train, x_test = x_train / 255.0, x_test / 255.0

print(f"Training data shape: {x_train.shape}")
print(f"Testing data shape: {x_test.shape}")
```

## 2. Feedforward Neural Network (FNN)

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense, Dropout

# Build FNN model
fnn_model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dropout(0.2),
    Dense(10, activation='softmax')
])

fnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("Training Feedforward Neural Network...")
history_fnn = fnn_model.fit(x_train, y_train, epochs=3, validation_split=0.1, verbose=1)

loss_fnn, acc_fnn = fnn_model.evaluate(x_test, y_test, verbose=0)
print(f"FNN Test Accuracy: {acc_fnn:.4f}")
```

## 3. Convolutional Neural Network (CNN)

```python
from tensorflow.keras.layers import Conv2D, MaxPooling2D

# Reshape data for CNN (add channel dimension)
x_train_cnn = x_train.reshape(-1, 28, 28, 1)
x_test_cnn = x_test.reshape(-1, 28, 28, 1)

# Build CNN model
cnn_model = Sequential([
    Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D(pool_size=(2, 2)),
    Conv2D(64, kernel_size=(3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("Training Convolutional Neural Network...")
history_cnn = cnn_model.fit(x_train_cnn, y_train, epochs=3, validation_split=0.1, verbose=1)

loss_cnn, acc_cnn = cnn_model.evaluate(x_test_cnn, y_test, verbose=0)
print(f"CNN Test Accuracy: {acc_cnn:.4f}")
```

## 4. Plot Training Histories

```python
plt.figure(figsize=(12, 5))

# Plot Accuracy
plt.subplot(1, 2, 1)
plt.plot(history_fnn.history['val_accuracy'], label='FNN Val Acc')
plt.plot(history_cnn.history['val_accuracy'], label='CNN Val Acc')
plt.title('Validation Accuracy Comparison')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

# Plot Loss
plt.subplot(1, 2, 2)
plt.plot(history_fnn.history['val_loss'], label='FNN Val Loss')
plt.plot(history_cnn.history['val_loss'], label='CNN Val Loss')
plt.title('Validation Loss Comparison')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.savefig('outputs/nn_comparison.png')
plt.close()
print("Plots saved to outputs/nn_comparison.png")
```
